```python
import sys
import platform
import json
import importlib.metadata as metadata
from pathlib import Path

packages = [
    "numpy",
    "scipy",
    "pandas",
    "matplotlib",
    "soundfile",
    "pesq",
    "tqdm",
    "ipywidgets",
    "ipython",
]

environment = {
    "colab_runtime": "2026.10",
    "python_version": platform.python_version(),
    "python_full_version": sys.version,
    "platform": platform.platform(),
    "packages": {}
}

for package in packages:
    try:
        environment["packages"][package] = metadata.version(package)
    except metadata.PackageNotFoundError:
        environment["packages"][package] = None

environment["validation"] = {
    "train_files": 1296,
    "validation_files": 164,
    "baseline_evaluation": "completed",
    "evaluation_notebook": "02_baseline_comparison.ipynb"
}

Path("/content/environment.json").write_text(
    json.dumps(environment, indent=2, ensure_ascii=False),
    encoding="utf-8"
)

print(json.dumps(environment, indent=2, ensure_ascii=False))
```

```python
import zipfile
from pathlib import Path

ZIP_PATH = Path(
    '/content/drive/MyDrive/signals_and_systems/'
    'assets/noise_reduction_challenge.zip'
)

EXTRACT_DIR = Path('/content')

with zipfile.ZipFile(ZIP_PATH, 'r') as z:
    z.extractall(EXTRACT_DIR)

DATA_ROOT = EXTRACT_DIR / 'noise_reduction_challenge'

assert (DATA_ROOT / 'train' / 'samples.csv').exists()
assert (DATA_ROOT / 'validation' / 'samples.csv').exists()
assert (DATA_ROOT / 'figures').exists()

print("Dataset preparation complete!")
print("DATA_ROOT:", DATA_ROOT)
```

```python
requirements = []

for package in packages:
    version = environment["packages"][package]

    if version is not None:
        requirements.append(f"{package}=={version}")

requirements_path = Path("/content/requirements.txt")

requirements_path.write_text(
    "\n".join(requirements) + "\n",
    encoding="utf-8"
)

print(requirements_path.read_text())
```
